```python
import numpy as np 
np.random.seed() 
```

```python
#Question 1 : Array Creation, Reshaping, and inspection 
a=np.arange(5,101,5) 
matrix=a.reshape(4,5) 

print("Matrix:\n", matrix) 
print("ndim:", matrix.ndim) 
print("shape:",matrix.shape) 
print("size:", matrix.size) 
print("dtype:", matrix.dtype) 
```

```python
#Question 2: Matrix slicing and sub-array modification 
b=np.ones((5,5), dtype=int) 
b[1:4,1:4]=0
b[2,2]=9 
print(b)
```

```python
#Question 3: Element-wise arthimetic and aggregates 
prices=np.array([120,250,80,450,300])
quantities = np.array([4,2,10,1,3]) 

revenue = prices * quantities
total = revenue.sum()
average = revenue.mean()
best=np.argmax(revenue)

print("Revenue per product:", revenue) 
print("Total revenue:", total) 
print("Average revenue:", average)
print("Highest earning product index:", best) 
```

```python
#Boolean masking and cleaning 
rainfall = np.array([12.5, -1.0, 0.0, 18.2, -99.0, 5.4, 22.1, 0.0, -1.0, 14.8])
valid = rainfall[rainfall >= 0] 
print("Valid:", valid) 

rainfall[rainfall < 0] = 0.0 
print("Cleaned:", rainfall) 

heavy_days= np.sum(rainfall >10.0) 
print("Heavy rainfall days:", heavy_days) 

```

```python
#Question 5: np.where() 
np.random.seed(10) 
scores=np.random.randint(20,101,size=12) 
print("Scores:", scores) 

status = np.where(scores >=40, "Pass", "Fail") 
print("Status:", status) 

updated_scores = np.where(scores < 40, scores+ 5, scores) 
print("Updated:", updated_scores)
```

```python
#Question 6:  axis-based aggregations 
np.random.seed(10) 
marks = np.random.randint(50,101,size=(6,4)) 

student_avg=marks.mean(axis=1) 
print("Student averages:", np.round(student_avg,2)) 

subject_max=marks.max(axis=0) 
print("Subject highs:", subject_max) 

print("Students with avg > 75:\n", marks[student_avg > 75]) 
```

```python
houses = np.array([[1200, 2, 55],
                   [1800, 3, 85],
                   [950,  1, 40],
                   [2400, 4, 120],
                   [1500, 3, 70]])

mins = houses.min(axis=0)      # per-column minimum -> shape (3,)
maxs = houses.max(axis=0)      # per-column maximum -> shape (3,)

normalized = (houses - mins) / (maxs - mins)   # broadcasting
print(np.round(normalized, 3))
```

```python
# Q8: Stacking 
morning_sales = np.array([
[15, 20],
[10, 12],
[25, 30]
])
evening_sales = np.array([
[18, 22],
[14, 15],
[20, 28]
])
full_day = np.hstack((morning_sales, evening_sales)) 
full_day = np.vstack((full_day, [12,16,15,19]))
print(full_day) 
print("Column totals:", full_day.sum(axis=0))
```

```python
#Question 9: Unique Elements and Frequency Distribution 
ratings = np.array([5,4,3,5,2,5,4,1,3,5,4,4,2,5,3,4,5,1,4,3])
values, counts = np.unique(ratings, return_counts=True) 
percent= counts/ratings.size*100 

for v,c,p in zip(values, counts, percent): 
    print(f"Rating {v}: {c} responses ({p:.1f}%)") 
    
```

```python
#Question10: argsort and top-k 
emp_ids = np.array(["E101","E102","E103","E104","E105","E106"])
scores = np.array([78,92,65,88,95,71]) 
order = np.argsort(scores)[::-1] 
sorted_ids = emp_ids[order] 
sorted_scores=scores[order] 
print("Top 3 employee:") 
print(sorted_ids[:3]) 
print(sorted_scores[:3]) 
```

```python
#1d array of shaping 
import numpy as np 
marks = np.array([1,2,3,4,5,6,7,8]) 
print(marks) 
print(marks.shape)
```

```python
#2d array of shaping 
import numpy as np 
marks = np.array([[80,90,75],[85,95,88]]) 
print(marks) 
print(marks.shape)
```

```python
import numpy as np 
a=[[1,2],[3,4]] 
b=[[5,6],[7,8]]
np.concatenate((a,b), axis=1)
```

```python
import numpy as np 
a=np.array([1,2,3,4])
b=np.array([5,6,7,8]) 
c= np.stack((a,b), axis=1)
print(c)
```

```python
import numpy as np 
a=np.array([1,2,3,4])
b=np.array([5,6,7,8]) 
c= np.hstack((a,b))
print(c)
```

```python
import numpy as np 
a=np.array([1,2,3,4])
b=np.array([5,6,7,8]) 
c= np.vstack((a,b))
print(c)
```

```python
import numpy as np

a = np.array([1, 2, 3, 4])
b = a.copy()

a[0] = 10

print(a)
print(b)
```

```python
import numpy as np

a = np.array([1, 2, 3, 4])
b = a.view()

a[0] = 10

print(a)
print(b)
```
